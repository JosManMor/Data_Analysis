## **5. Función de conversión**

Crea una función para convertir los valores de la temperatura del dataset que se encuentran en grados Celsius a grados Fahrenheit y agrega la conversión en una nueva columna.

```python
# Definimos la función que convierte de Celsius a Fahrenheit

def celsius_to_farenheit(c):
  return c * 9 / 5 + 32
```

```python
# Definimos una nueva columna donde se guardará la conversión
weather_ds['Temperature (F)'] = weather_ds['Temperature'].apply(celsius_to_farenheit)
```

## **10. Escalamiento y normalización.**

Haz el escalamiento de las características temperatura, humedad, velocidad del viento y precipitación. Escala las características presión atmosférica, UV index y visibilidad utilizando la normalización.

```python
# Cargamos la librería preprocessing para el escalamiento de datos
from sklearn import preprocessing
```

```python
# Mostramos los valores estadísticos para verificar los valores
weather_ds.describe()
```

```python
escalar = ['Temperature', 'Humidity', 'Wind Speed', 'Precipitation (%)']
normalizar = ['Atmospheric Pressure', 'UV Index', 'Visibility (km)']

weather_scaled = weather_ds.copy()

# Escalamiento Min-Max
weather_scaled[escalar] = (weather_ds[escalar] - weather_ds[escalar].min()) / \
                          (weather_ds[escalar].max() - weather_ds[escalar].min())

# Normalización Z-score
weather_scaled[normalizar] = (weather_ds[normalizar] - weather_ds[normalizar].mean()) / \
                             weather_ds[normalizar].std()

weather_scaled[escalar + normalizar].describe()
```

## **12. Gráficas de burbuja.**

Crea un gráfico de burbuja definiendo una característica para el uso de colores.

```python
# Muestra para que las burbujas no se encimen (el dataset tiene miles de filas)
muestra = weather_ds.sample(300, random_state=42)

sns.set_theme(style='darkgrid')
plt.figure(figsize=(12, 7))
sns.scatterplot(
    data=muestra,
    x='Temperature',
    y='Humidity',
    size='Wind Speed',
    sizes=(20, 400),
    alpha=0.5,
    hue='Weather Type'
)
plt.title('Gráfico tipo bubble chart con manejo de colores')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.show()
```

## **13. Gráfico box plot.**

Crea un gráfico de bigotes para representar la información del dataset.

```python
sns.set_palette('Paired')
ax = sns.boxplot(x='Weather Type', y='Temperature', hue='Weather Type', data=weather_ds)
ax.set_xlabel('Tipo de clima')
ax.set_ylabel('Temperatura [°C]')
ax.set_title('Distribución de la Temperatura por Tipo de Clima')
plt.show()
```

## **14. Gráfico de pila.**

Crea un gráfico stacked para representar datos del dataset.

```python
# Agrupamos los datos por Cloud cover y weather type.
cloud_weather_counts = weather_ds.groupby(['Cloud Cover', 'Weather Type']).size().unstack(fill_value=0)
display(cloud_weather_counts.head())
```

```python
# Gráfico de Barras apiladas de ocurrencias por tipo de clima y tipo de nubes
cloud_weather_counts.plot(kind='bar', stacked=True, figsize=(12, 7))
plt.title('Distribución de Tipos de Clima por Tipo de Nubes')
plt.xlabel('Tipo de Nubes')
plt.ylabel('Cantidad de Ocurrencias')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Tipo de Clima', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
