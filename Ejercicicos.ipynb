```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
hr_df["Age"] = scaler.fit_transform(hr_df["Age"].values.reshape(-1,1))
```

```python
hr_df["Age"].describe()
```

```python
# La estandarizacion transforma los datos para que tengan una media de cero y una desviacion estandar de 1
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
# se coloca un -1 en el reshape para indicar un numero indefinido de columnas
hr_df["MonthlyIncome"] = scaler.fit_transform(hr_df["MonthlyIncome"].values.reshape(-1,1))
```

```python
hr_df["MonthlyIncome"].describe().round(2)
```
